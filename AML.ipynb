```python
import os
import cv2
import numpy as np
import torch
from torch.utils.data import Dataset
import albumentations as A
from albumentations.pytorch import ToTensorV2
```

```python
data_path = "/content/drive/MyDrive/aml_segmentation/data/wbcatt_plus/pbcseg_final_v1/pbcseg_final_v1"

sample_files = [f for f in os.listdir(data_path) if f.endswith('_ccrop_mask.png')]
sample_mask_path = os.path.join(data_path, sample_files[0])

mask = cv2.imread(sample_mask_path, cv2.IMREAD_UNCHANGED)
print("Mask file:", sample_files[0])
print("Mask shape:", mask.shape)
print("Mask dtype:", mask.dtype)

if len(mask.shape) == 2:
    unique_vals = np.unique(mask)
    print("Unique pixel values (grayscale mask):", unique_vals)
else:
    pixels = mask.reshape(-1, mask.shape[-1])
    unique_colors = np.unique(pixels, axis=0)
    print("Unique colors found (up to 20 shown):")
    print(unique_colors[:20])
```

```python
def mask_to_classes(mask):
    out = np.zeros_like(mask, dtype=np.uint8)
    out[mask == 1] = 1   # cytoplasm
    out[mask == 5] = 1   # vacuole -> cytoplasm
    out[mask == 2] = 2   # nucleus
    return out
```

```python
IMG_SIZE = 256

train_aug = A.Compose([
    A.Resize(IMG_SIZE, IMG_SIZE),
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.5),
    A.Rotate(limit=15, p=0.7, border_mode=cv2.BORDER_CONSTANT, value=0, mask_value=0),
    A.Affine(translate_percent=0.05, scale=(0.95, 1.05), rotate=0, p=0.5),
    A.ColorJitter(brightness=0.15, contrast=0.15, saturation=0.1, hue=0.02, p=0.5),
    A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2(),
])

val_aug = A.Compose([
    A.Resize(IMG_SIZE, IMG_SIZE),
    A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2(),
])
```

```python
class WBCData(Dataset):
    def __init__(self, folder, ids, aug=None):
        self.folder = folder
        self.ids = ids
        self.aug = aug

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, i):
        name = self.ids[i]
        img_path = os.path.join(self.folder, name + "_ccrop.jpg")
        mask_path = os.path.join(self.folder, name + "_ccrop_mask.png")

        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        raw_mask = cv2.imread(mask_path, cv2.IMREAD_UNCHANGED)
        mask = mask_to_classes(raw_mask)

        if self.aug:
            result = self.aug(image=img, mask=mask)
            img = result['image']
            mask = result['mask'].long()

        return img, mask
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import segmentation_models_pytorch as smp
```

```python
# primary backbone - Attention U-Net

def build_attention_unet(num_classes=3):
    model = smp.Unet(
        encoder_name="resnet34",
        encoder_weights="imagenet",
        decoder_attention_type="scse",
        in_channels=3,
        classes=num_classes,
    )
    return model

model = build_attention_unet(num_classes=3)
```

**Ratio-Consistency Loss**

```python
class RatioConsistencyLoss(nn.Module):

    def __init__(self, epsilon=1e-6):
        super().__init__()
        self.epsilon = epsilon
        self.huber = nn.SmoothL1Loss()  # Huber loss, same as paper's Eq. 5

    def forward(self, pred_probs, true_mask):
        # pred_probs: [batch, num_classes, H, W] - softmax probabilities
        # true_mask: [batch, H, W] - integer class labels (0=bg, 1=cyto, 2=nucleus)

        pred_nucleus_prob = pred_probs[:, 2, :, :]   # soft nucleus prediction
        pred_cyto_prob = pred_probs[:, 1, :, :]      # soft cytoplasm prediction

        pred_nucleus_area = pred_nucleus_prob.sum(dim=[1, 2])
        pred_cyto_area = pred_cyto_prob.sum(dim=[1, 2])

        # real areas from the ground truth mask
        true_nucleus_area = (true_mask == 2).float().sum(dim=[1, 2])
        true_cyto_area = (true_mask == 1).float().sum(dim=[1, 2])

        pred_ratio = torch.log((pred_nucleus_area + self.epsilon) / (pred_cyto_area + self.epsilon))
        true_ratio = torch.log((true_nucleus_area + self.epsilon) / (true_cyto_area + self.epsilon))

        return self.huber(pred_ratio, true_ratio)
```

**Total Loss**

```python
class TotalLoss(nn.Module):
    """
    Combines the standard segmentation losses with the new ratio loss.
    L = L_Dice + lambda1 * L_Focal + lambda2 * L_ratio
    """

    def __init__(self, lambda1=0.5, lambda2=0.3):
        super().__init__()
        self.dice_loss = smp.losses.DiceLoss(mode='multiclass')
        self.focal_loss = smp.losses.FocalLoss(mode='multiclass')
        self.ratio_loss = RatioConsistencyLoss()
        self.lambda1 = lambda1
        self.lambda2 = lambda2

    def forward(self, pred_logits, true_mask):
        pred_probs = F.softmax(pred_logits, dim=1)

        dice = self.dice_loss(pred_logits, true_mask)
        focal = self.focal_loss(pred_logits, true_mask)
        ratio = self.ratio_loss(pred_probs, true_mask)

        total = dice + self.lambda1 * focal + self.lambda2 * ratio

        return total, {
            "dice": dice.item(),
            "focal": focal.item(),
            "ratio": ratio.item(),
            "total": total.item(),
        }
```

```python
model = model.cuda() if torch.cuda.is_available() else model
loss_fn = TotalLoss(lambda1=0.5, lambda2=0.3)

imgs, masks = next(iter(train_loader))
if torch.cuda.is_available():
    imgs = imgs.cuda()
    masks = masks.cuda()

preds = model(imgs)
loss_value, loss_parts = loss_fn(preds, masks)

print("total loss:", loss_value.item())
print("breakdown:", loss_parts)
```

**Optimizer and scheduler setup**

```python
import torch.optim as optim

LEARNING_RATE = 1e-4
WEIGHT_DECAY = 1e-4
MAX_EPOCHS = 100
PATIENCE = 15

optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)

scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=MAX_EPOCHS)
```

```python
def dice_score(pred_logits, true_mask, num_classes=3, smooth=1e-6):
    pred_classes = torch.argmax(pred_logits, dim=1)

    dice_per_class = []
    for c in range(1, num_classes):
        pred_c = (pred_classes == c).float()
        true_c = (true_mask == c).float()

        intersection = (pred_c * true_c).sum()
        union = pred_c.sum() + true_c.sum()

        dice = (2 * intersection + smooth) / (union + smooth)
        dice_per_class.append(dice.item())

    return sum(dice_per_class) / len(dice_per_class)
```

**N:C ratio error function**

```python
def nc_ratio_error(pred_logits, true_mask, epsilon=1e-6):
    pred_classes = torch.argmax(pred_logits, dim=1)

    pred_nucleus = (pred_classes == 2).float().sum(dim=[1, 2])
    pred_cyto = (pred_classes == 1).float().sum(dim=[1, 2])
    true_nucleus = (true_mask == 2).float().sum(dim=[1, 2])
    true_cyto = (true_mask == 1).float().sum(dim=[1, 2])

    pred_ratio = pred_nucleus / (pred_cyto + epsilon)
    true_ratio = true_nucleus / (true_cyto + epsilon)

    error = torch.abs(pred_ratio - true_ratio)
    return error.mean().item()
```

```python
def train_one_epoch(model, loader, loss_fn, optimizer, device):
    model.train()
    running_loss = 0.0

    for imgs, masks in loader:
        imgs = imgs.to(device)
        masks = masks.to(device)

        optimizer.zero_grad()
        preds = model(imgs)
        loss, _ = loss_fn(preds, masks)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    return running_loss / len(loader)
```

```python
def validate_one_epoch(model, loader, loss_fn, device):
    model.eval()
    running_loss = 0.0
    running_dice = 0.0
    running_ratio_error = 0.0

    with torch.no_grad():
        for imgs, masks in loader:
            imgs = imgs.to(device)
            masks = masks.to(device)

            preds = model(imgs)
            loss, _ = loss_fn(preds, masks)

            running_loss += loss.item()
            running_dice += dice_score(preds, masks)
            running_ratio_error += nc_ratio_error(preds, masks)

    n = len(loader)
    return running_loss / n, running_dice / n, running_ratio_error / n
```

```python
import random
import numpy as np

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
```

```python
import os

def train_model(model, train_loader, val_loader, loss_fn, optimizer, scheduler,
                 device, max_epochs, patience, checkpoint_path, seed=42):

    set_seed(seed)
    model = model.to(device)

    best_dice = -1
    best_ratio_error = float('inf')
    epochs_no_improve = 0

    history = []

    for epoch in range(1, max_epochs + 1):
        train_loss = train_one_epoch(model, train_loader, loss_fn, optimizer, device)
        val_loss, val_dice, val_ratio_error = validate_one_epoch(model, val_loader, loss_fn, device)

        scheduler.step()

        print(f"Epoch {epoch}: train_loss={train_loss:.4f}  val_loss={val_loss:.4f}  "
              f"val_dice={val_dice:.4f}  val_ratio_error={val_ratio_error:.4f}")

        history.append({
            "epoch": epoch,
            "train_loss": train_loss,
            "val_loss": val_loss,
            "val_dice": val_dice,
            "val_ratio_error": val_ratio_error,
        })

        # main rule: best mean dice, tie-breaker: lower ratio error
        is_better = (val_dice > best_dice) or (val_dice == best_dice and val_ratio_error < best_ratio_error)

        if is_better:
            best_dice = val_dice
            best_ratio_error = val_ratio_error
            epochs_no_improve = 0
            torch.save(model.state_dict(), checkpoint_path)
            print(f"  -> new best model saved (dice={best_dice:.4f})")
        else:
            epochs_no_improve += 1

        if epochs_no_improve >= patience:
            print(f"Stopping early at epoch {epoch}, no improvement for {patience} epochs.")
            break

    return history
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

checkpoint_path = "/content/drive/MyDrive/aml_segmentation/checkpoints/attention_unet_seed42.pth"
os.makedirs(os.path.dirname(checkpoint_path), exist_ok=True)

history = train_model(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    loss_fn=loss_fn,
    optimizer=optimizer,
    scheduler=scheduler,
    device=device,
    max_epochs=MAX_EPOCHS,
    patience=PATIENCE,
    checkpoint_path=checkpoint_path,
    seed=42,
)
```
