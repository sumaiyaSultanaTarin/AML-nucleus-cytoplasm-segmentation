```python
def mask_to_classes(mask):
    out = np.zeros_like(mask, dtype=np.uint8)
    out[mask == 1] = 1   # cytoplasm
    out[mask == 5] = 1   # vacuole -> cytoplasm
    out[mask == 2] = 2   # nucleus
    return out
```

The actual WBCData Dataset class

```python
class WBCData(Dataset):
    def __init__(self, folder, ids, aug=None):
        self.folder = folder
        self.ids = ids
        self.aug = aug

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, i):
        name = self.ids[i]
        img_path = os.path.join(self.folder, name + "_ccrop.jpg")
        mask_path = os.path.join(self.folder, name + "_ccrop_mask.png")

        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        raw_mask = cv2.imread(mask_path, cv2.IMREAD_UNCHANGED)
        mask = mask_to_classes(raw_mask)

        if self.aug:
            result = self.aug(image=img, mask=mask)
            img = result['image']
            mask = result['mask'].long()

        return img, mask
```

**Section 8: Build the Custom Loss**

**Ratio-Consistency Loss**

```python
class RatioConsistencyLoss(nn.Module):

    def __init__(self, epsilon=1e-6):
        super().__init__()
        self.epsilon = epsilon
        self.huber = nn.SmoothL1Loss()  # Huber loss

    def forward(self, pred_probs, true_mask):
        # pred_probs: [batch, num_classes, H, W] - softmax probabilities
        # true_mask: [batch, H, W] - integer class labels (0=bg, 1=cyto, 2=nucleus)

        pred_nucleus_prob = pred_probs[:, 2, :, :]   # soft nucleus prediction
        pred_cyto_prob = pred_probs[:, 1, :, :]      # soft cytoplasm prediction

        pred_nucleus_area = pred_nucleus_prob.sum(dim=[1, 2])
        pred_cyto_area = pred_cyto_prob.sum(dim=[1, 2])

        # real areas from the ground truth mask
        true_nucleus_area = (true_mask == 2).float().sum(dim=[1, 2])
        true_cyto_area = (true_mask == 1).float().sum(dim=[1, 2])

        pred_ratio = torch.log((pred_nucleus_area + self.epsilon) / (pred_cyto_area + self.epsilon))
        true_ratio = torch.log((true_nucleus_area + self.epsilon) / (true_cyto_area + self.epsilon))

        return self.huber(pred_ratio, true_ratio)
```

**Total Loss**

```python
class TotalLoss(nn.Module):
    """
    Combines the standard segmentation losses with the new ratio loss.
    L = L_Dice + lambda1 * L_Focal + lambda2 * L_ratio
    """

    def __init__(self, lambda1=0.5, lambda2=0.3):
        super().__init__()
        self.dice_loss = smp.losses.DiceLoss(mode='multiclass')
        self.focal_loss = smp.losses.FocalLoss(mode='multiclass')
        self.ratio_loss = RatioConsistencyLoss()
        self.lambda1 = lambda1
        self.lambda2 = lambda2

    def forward(self, pred_logits, true_mask):
        pred_probs = F.softmax(pred_logits, dim=1)

        dice = self.dice_loss(pred_logits, true_mask)
        focal = self.focal_loss(pred_logits, true_mask)
        ratio = self.ratio_loss(pred_probs, true_mask)

        total = dice + self.lambda1 * focal + self.lambda2 * ratio

        return total, {
            "dice": dice.item(),
            "focal": focal.item(),
            "ratio": ratio.item(),
            "total": total.item(),
        }
```

**Optimizer and scheduler setup**

```python
import torch.optim as optim

LEARNING_RATE = 1e-4
WEIGHT_DECAY = 1e-4
MAX_EPOCHS = 100
PATIENCE = 15

optimizer = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)

scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=MAX_EPOCHS)
```

```python
def dice_score(pred_logits, true_mask, num_classes=3, smooth=1e-6):
    pred_classes = torch.argmax(pred_logits, dim=1)

    dice_per_class = []
    for c in range(1, num_classes):
        pred_c = (pred_classes == c).float()
        true_c = (true_mask == c).float()

        intersection = (pred_c * true_c).sum()
        union = pred_c.sum() + true_c.sum()

        dice = (2 * intersection + smooth) / (union + smooth)
        dice_per_class.append(dice.item())

    return sum(dice_per_class) / len(dice_per_class)
```

**N:C ratio error function**

```python
def nc_ratio_error(pred_logits, true_mask, epsilon=1e-6):
    pred_classes = torch.argmax(pred_logits, dim=1)

    pred_nucleus = (pred_classes == 2).float().sum(dim=[1, 2])
    pred_cyto = (pred_classes == 1).float().sum(dim=[1, 2])
    true_nucleus = (true_mask == 2).float().sum(dim=[1, 2])
    true_cyto = (true_mask == 1).float().sum(dim=[1, 2])

    pred_ratio = pred_nucleus / (pred_cyto + epsilon)
    true_ratio = true_nucleus / (true_cyto + epsilon)

    error = torch.abs(pred_ratio - true_ratio)
    return error.mean().item()
```

```python
def train_one_epoch(model, loader, loss_fn, optimizer, device):
    model.train()
    running_loss = 0.0

    for imgs, masks in loader:
        imgs = imgs.to(device)
        masks = masks.to(device)

        optimizer.zero_grad()
        preds = model(imgs)
        loss, _ = loss_fn(preds, masks)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()

    return running_loss / len(loader)
```

```python
def validate_one_epoch(model, loader, loss_fn, device):
    model.eval()
    running_loss = 0.0
    running_dice = 0.0
    running_ratio_error = 0.0

    with torch.no_grad():
        for imgs, masks in loader:
            imgs = imgs.to(device)
            masks = masks.to(device)

            preds = model(imgs)
            loss, _ = loss_fn(preds, masks)

            running_loss += loss.item()
            running_dice += dice_score(preds, masks)
            running_ratio_error += nc_ratio_error(preds, masks)

    n = len(loader)
    return running_loss / n, running_dice / n, running_ratio_error / n
```

```python
import random
import numpy as np

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
```

```python
import os

def train_model(model, train_loader, val_loader, loss_fn, optimizer, scheduler,
                 device, max_epochs, patience, checkpoint_path, seed=42):

    set_seed(seed)
    model = model.to(device)

    best_dice = -1
    best_ratio_error = float('inf')
    epochs_no_improve = 0

    history = []

    for epoch in range(1, max_epochs + 1):
        train_loss = train_one_epoch(model, train_loader, loss_fn, optimizer, device)
        val_loss, val_dice, val_ratio_error = validate_one_epoch(model, val_loader, loss_fn, device)

        scheduler.step()

        print(f"Epoch {epoch}: train_loss={train_loss:.4f}  val_loss={val_loss:.4f}  "
              f"val_dice={val_dice:.4f}  val_ratio_error={val_ratio_error:.4f}")

        history.append({
            "epoch": epoch,
            "train_loss": train_loss,
            "val_loss": val_loss,
            "val_dice": val_dice,
            "val_ratio_error": val_ratio_error,
        })

        is_better = (val_dice > best_dice) or (val_dice == best_dice and val_ratio_error < best_ratio_error)

        if is_better:
            best_dice = val_dice
            best_ratio_error = val_ratio_error
            epochs_no_improve = 0
            torch.save(model.state_dict(), checkpoint_path)
            print(f"  -> new best model saved (dice={best_dice:.4f})")
        else:
            epochs_no_improve += 1

        if epochs_no_improve >= patience:
            print(f"Stopping early at epoch {epoch}, no improvement for {patience} epochs.")
            break

    return history
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Using device:", device)

checkpoint_path = "/content/drive/MyDrive/aml_segmentation/checkpoints/attention_unet_seed42.pth"
import os
os.makedirs(os.path.dirname(checkpoint_path), exist_ok=True)

model = model.to(device)
best_dice = -1
best_ratio_error = float('inf')
epochs_no_improve = 0

set_seed(42)

for epoch in range(1, MAX_EPOCHS + 1):
    model.train()
    running_loss = 0.0
    batch_count = 0

    for imgs, masks in train_loader:
        imgs = imgs.to(device)
        masks = masks.to(device)

        optimizer.zero_grad()
        preds = model(imgs)
        loss, parts = loss_fn(preds, masks)
        loss.backward()
        optimizer.step()

        running_loss += loss.item()
        batch_count += 1

        if batch_count % 10 == 0:
            print(f"  epoch {epoch}, batch {batch_count}/{len(train_loader)}, loss so far: {running_loss/batch_count:.4f}")

    train_loss = running_loss / batch_count

    model.eval()
    val_loss = 0.0
    val_dice = 0.0
    val_ratio_error = 0.0
    val_batches = 0

    with torch.no_grad():
        for imgs, masks in val_loader:
            imgs = imgs.to(device)
            masks = masks.to(device)
            preds = model(imgs)
            loss, _ = loss_fn(preds, masks)

            val_loss += loss.item()
            val_dice += dice_score(preds, masks)
            val_ratio_error += nc_ratio_error(preds, masks)
            val_batches += 1

    val_loss /= val_batches
    val_dice /= val_batches
    val_ratio_error /= val_batches

    scheduler.step()

    print(f"EPOCH {epoch} DONE -> train_loss={train_loss:.4f} val_loss={val_loss:.4f} val_dice={val_dice:.4f} val_ratio_error={val_ratio_error:.4f}")

    torch.save(model.state_dict(), checkpoint_path.replace(".pth", "_latest.pth"))

    is_better = (val_dice > best_dice) or (val_dice == best_dice and val_ratio_error < best_ratio_error)
    if is_better:
        best_dice = val_dice
        best_ratio_error = val_ratio_error
        epochs_no_improve = 0
        torch.save(model.state_dict(), checkpoint_path)
        print(f"  -> NEW BEST saved (dice={best_dice:.4f})")
    else:
        epochs_no_improve += 1

    if epochs_no_improve >= PATIENCE:
        print(f"Stopping early at epoch {epoch}, no improvement for {PATIENCE} epochs.")
        break
```

```python
import os
import random

base_path = "/content/drive/MyDrive/PKG - AML-Cytomorphology_LMU"

blast_lineage_classes = ['MYO', 'MOB', 'PMO', 'PMB', 'MYB', 'MMZ']
all_classes = [f for f in os.listdir(base_path) if os.path.isdir(os.path.join(base_path, f))]
control_classes = [c for c in all_classes if c not in blast_lineage_classes]

TARGET_TOTAL = 450
TARGET_BLAST = int(TARGET_TOTAL * 0.6)   # 270
TARGET_CONTROL = TARGET_TOTAL - TARGET_BLAST  # 180

random.seed(42)

def smart_sample(class_list, n_total, base_path):
    available = {}
    for cls in class_list:
        cls_path = os.path.join(base_path, cls)
        available[cls] = os.listdir(cls_path)

    selected = []
    remaining_need = n_total
    classes_left = list(class_list)

    # give small/rare classes everything they have first,
    # then split the leftover need across the remaining bigger classes
    while remaining_need > 0 and classes_left:
        share = remaining_need // len(classes_left)
        if share == 0:
            share = 1

        next_round_classes = []
        for cls in classes_left:
            files = available[cls]
            take = min(share, len(files))
            chosen = random.sample(files, take)
            for f in chosen:
                selected.append((cls, f))
            remaining_need -= take
            available[cls] = [f for f in files if f not in chosen]

            if available[cls]:
                next_round_classes.append(cls)

        classes_left = next_round_classes
        if remaining_need <= 0:
            break

    return selected

blast_selection = smart_sample(blast_lineage_classes, TARGET_BLAST, base_path)
control_selection = smart_sample(control_classes, TARGET_CONTROL, base_path)

full_subset = blast_selection + control_selection
random.shuffle(full_subset)

print("Blast-lineage selected:", len(blast_selection))
print("Control/mature selected:", len(control_selection))
print("Total selected:", len(full_subset))

from collections import Counter
print("\nBlast breakdown:", Counter([c for c, f in blast_selection]))
print("Control breakdown:", Counter([c for c, f in control_selection]))
```

```python
class AMLPredMaskDataset(Dataset):
    def __init__(self, image_folder, mask_folder, filenames, aug=None):
        self.image_folder = image_folder
        self.mask_folder = mask_folder
        self.filenames = filenames
        self.aug = aug

    def __len__(self):
        return len(self.filenames)

    def __getitem__(self, i):
        filename = self.filenames[i]
        img_path = os.path.join(self.image_folder, filename)
        mask_filename = filename.replace('.tiff', '_predmask.png')
        mask_path = os.path.join(self.mask_folder, mask_filename)

        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = cv2.resize(img, (256, 256))  # match the mask's resolution

        mask = cv2.imread(mask_path, cv2.IMREAD_UNCHANGED)

        if self.aug:
            result = self.aug(image=img, mask=mask)
            img = result['image']
            mask = result['mask'].long()

        return img, mask
```

```python
# start from the WBCAtt+ trained checkpoint, then fine-tune on AML-LMU
transfer_model = make_model("attention_unet").to(device)
transfer_model.load_state_dict(torch.load("/content/drive/MyDrive/aml_segmentation/checkpoints/attention_unet_seed42.pth"))

transfer_optimizer = optim.AdamW(transfer_model.parameters(), lr=LEARNING_RATE, weight_decay=WEIGHT_DECAY)
transfer_scheduler = optim.lr_scheduler.CosineAnnealingLR(transfer_optimizer, T_max=MAX_EPOCHS)
transfer_loss_fn = TotalLoss(lambda1=0.5, lambda2=0.3)

transfer_checkpoint_path = "/content/drive/MyDrive/aml_segmentation/checkpoints/attention_unet_transfer_aml_seed42.pth"
os.makedirs(os.path.dirname(transfer_checkpoint_path), exist_ok=True)

set_seed(42)

best_dice = -1
best_ratio_error = float('inf')
epochs_no_improve = 0

for epoch in range(1, MAX_EPOCHS + 1):
    transfer_model.train()
    running_loss = 0.0
    batch_count = 0

    for imgs, masks in aml_train_loader:
        imgs = imgs.to(device)
        masks = masks.to(device)

        transfer_optimizer.zero_grad()
        preds = transfer_model(imgs)
        loss, _ = transfer_loss_fn(preds, masks)
        loss.backward()
        transfer_optimizer.step()

        running_loss += loss.item()
        batch_count += 1

    train_loss = running_loss / batch_count

    transfer_model.eval()
    val_loss = 0.0
    val_dice = 0.0
    val_ratio_error = 0.0
    val_batches = 0

    with torch.no_grad():
        for imgs, masks in aml_val_loader:
            imgs = imgs.to(device)
            masks = masks.to(device)
            preds = transfer_model(imgs)
            loss, _ = transfer_loss_fn(preds, masks)

            val_loss += loss.item()
            val_dice += dice_score(preds, masks)
            val_ratio_error += nc_ratio_error(preds, masks)
            val_batches += 1

    val_loss /= val_batches
    val_dice /= val_batches
    val_ratio_error /= val_batches

    transfer_scheduler.step()

    print(f"EPOCH {epoch} DONE -> train_loss={train_loss:.4f} val_loss={val_loss:.4f} val_dice={val_dice:.4f} val_ratio_error={val_ratio_error:.4f}")

    is_better = (val_dice > best_dice) or (val_dice == best_dice and val_ratio_error < best_ratio_error)
    if is_better:
        best_dice = val_dice
        best_ratio_error = val_ratio_error
        epochs_no_improve = 0
        torch.save(transfer_model.state_dict(), transfer_checkpoint_path)
        print(f"  -> NEW BEST saved (dice={best_dice:.4f})")
    else:
        epochs_no_improve += 1

    if epochs_no_improve >= PATIENCE:
        print(f"Stopping early at epoch {epoch}, no improvement for {PATIENCE} epochs.")
        break
```
